# 02 - Data Cleaning

## Objective

The objective of this notebook is to transform the raw Superstore dataset into a clean, analysis-ready dataset suitable for SQL analysis and Power BI dashboards.

## Tasks Performed

- Standardized column names
- Converted date columns to datetime
- Validated numeric fields
- Checked missing values
- Checked duplicate records
- Identified invalid records
- Performed outlier detection using IQR
- Exported cleaned dataset

```python
# Import libraries
import pandas as pd
import numpy as np
```

```python
# Load raw dataset
df = pd.read_csv("Superstore.csv", encoding="latin1")
```

```python
# Display first five rows
df.head()
```

```python
# Display Before cleaning column names
print(df.columns.tolist())
```

```python
# Standardize Column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ","_")
    .str.replace("-","_")
)
```

```python
# Display after cleaning column name
print(df.columns.tolist())
```

```python
# Display sample date values
df[['order_date','ship_date']].head(10)
```

```python
# convert data columns to datetime
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format = "mixed",
    dayfirst = True
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    format = "mixed",
    dayfirst = True
)
```

```python
df[["order_date", "ship_date"]].dtypes
```

```python
print(df["order_date"].isna().sum())
print(df["ship_date"].isna().sum())
```

```python
df.head()
```

```python
# summary statistics for numeric columns
df[["sales", "quantity","discount","profit"]].describe()
```

```python
# Orders with zero or nefative sales
invalid_sales = df[df["sales"] <= 0]

print(f"Invalid sales records: {len(invalid_sales)}")
```

```python
invalid_quantity = df[df["quantity"] <= 0]
print(f"Invalid Quantity Records: {len(invalid_quantity)}")
```

```python
invalid_discount = df[ (df["discount"] < 0) | (df["discount"] > 1)]

print(f"Invalid Discount Records: {len(invalid_discount)}")
```

```python
loss_orders = df[df["profit"] < 0]
print(f"Loss Making Orders: {len(loss_orders)}")
```

```python
def detect_outliers_iqr(data,column):
        # Detect outliers in a numeric column using the IQP method.

    # Calculate Q1 and Q3
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate lower and upper bounds
    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    # Identify outliers
    outliers = data[ (data[column] < lower_bound) | (data[column] > upper_bound)]

    print(f"{column}")
    print(f"Lower Bound : {lower_bound : 2f}")
    print(f"Upper Bound : {upper_bound : 2f}")
    print(f"Outliers : {len(outliers)}")

    return outliers
```

```python
sales_outliers = detect_outliers_iqr(df,"sales")
```

Outliers were detected using the Interquartile Range (IQR) method. A total of 1,167 sales transactions exceeded the upper bound of $498.93. These records were reviewed and retained because they represent legitimate high-value retail transactions rather than data entry errors. Removing them would distort revenue, customer, and product analyses.

```python
profit_outliers = detect_outliers_iqr(df, "profit")
```

Using the IQR method, 1,881 profit outliers were identified.

These records were retained because they represent genuine business scenarios, including highly profitable orders and loss-making transactions. Retaining these observations preserves the integrity of profitability analysis and supports business decision-making.

```python
# Save cleaned dataset
output_path = "data/processed/Superstore_Clean.csv"
df.to_csv(output_path, index=False)
print("✅ Cleaned dataset saved successfully!")
```
